```python
from tensorflow.keras.models import load_model, Model
import numpy as np

# Charger ton modèle fine-tuned
model = load_model("mobilenetv2_finetuned_best.keras")

# Vérifier les couches
print("Nombre de couches :", len(model.layers))

# Créer extracteur de features (avant classification)
feature_extractor = Model(
    inputs=model.input,
    outputs=model.layers[-2].output
)

print("Feature extractor prêt ✅")
print("Shape sortie :", feature_extractor.output_shape)
```

```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split

# chemins (adapte si besoin)
DATA_DIR = "faces_dataset"
CSV_FILE = "labels_corrected.csv"

# charger CSV
df = pd.read_csv(CSV_FILE)

# ajouter chemin complet
df["path"] = df["image_name"].apply(lambda x: os.path.join(DATA_DIR, x))

# label en string
df["label"] = df["label"].astype(str)

# split EXACTEMENT comme ton training
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["label"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=42
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
train_df.to_csv("train_split.csv", index=False)
test_df.to_csv("test_split.csv", index=False)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import tensorflow as tf

IMG_SIZE = 224
BATCH_SIZE = 32
DATA_DIR = "faces_dataset"

datagen = ImageDataGenerator(
    preprocessing_function=tf.keras.applications.mobilenet_v2.preprocess_input
)

train_generator = datagen.flow_from_dataframe(
    dataframe=train_df,
    directory=DATA_DIR,
    x_col="image_name",
    y_col="label",
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

test_generator = datagen.flow_from_dataframe(
    dataframe=test_df,
    directory=DATA_DIR,
    x_col="image_name",
    y_col="label",
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

print("Classes mapping :", train_generator.class_indices)
print("Train samples :", train_generator.samples)
print("Test samples :", test_generator.samples)
```

```python
print("Extraction features TRAIN...")
X_train_features = feature_extractor.predict(train_generator, verbose=1)

print("Extraction features TEST...")
X_test_features = feature_extractor.predict(test_generator, verbose=1)

y_train = train_generator.classes
y_test  = test_generator.classes

print("Train shape:", X_train_features.shape)
print("Test shape :", X_test_features.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_features = scaler.fit_transform(X_train_features)
X_test_features  = scaler.transform(X_test_features)

print("Normalisation OK ✅")
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

k = 80  # 🔥 paramètre important (on pourra optimiser)

selector = SelectKBest(score_func=f_classif, k=k)

X_train_sel = selector.fit_transform(X_train_features, y_train)
X_test_sel  = selector.transform(X_test_features)

print("Selected features shape:", X_train_sel.shape)
```

```python
import numpy as np

weights = selector.scores_
weights = weights / np.sum(weights)

print("Weights OK ✅")
```

```python
def build_fuzzy_stats(X_train_sel, y_train):
    classes = np.unique(y_train)
    stats = {}

    for c in classes:
        X_c = X_train_sel[y_train == c]
        stats[c] = {
            "mean": np.mean(X_c, axis=0),
            "std": np.std(X_c, axis=0) + 1e-6
        }

    return stats

stats = build_fuzzy_stats(X_train_sel, y_train)

print("Fuzzy stats ready ✅")
```

```python
def fuzzy_predict(X, stats, weights):
    preds = []

    for x in X:
        scores = {}

        for c in stats:
            mean = stats[c]["mean"]
            std  = stats[c]["std"]

            # fonction gaussienne
            membership = np.exp(-((x - mean)**2) / (2 * std**2))

            # pondération intelligente
            score = np.sum(weights * membership)

            scores[c] = score

        pred = max(scores, key=scores.get)
        preds.append(pred)

    return np.array(preds)
```

```python
from sklearn.metrics import accuracy_score

y_pred = fuzzy_predict(X_test_sel, stats, weights)

acc = accuracy_score(y_test, y_pred)

print("🔥 QLF Accuracy:", acc)
```

```python
def create_membership_functions(X, y):
    classes = np.unique(y)
    mf = {}

    for c in classes:
        X_c = X[y == c]
        mf[c] = {
            "mean": np.mean(X_c, axis=0),
            "std": np.std(X_c, axis=0)
        }
    return mf

mf = create_membership_functions(X_train_sel, y_train)
```

```python
def fuzzy_predict(X, mf, weights):
    preds = []

    for x in X:
        scores = {}

        for c in mf:
            mean = mf[c]["mean"]
            std  = mf[c]["std"] + 1e-6

            # gaussienne
            membership = np.exp(-((x - mean)**2) / (2 * std**2))

            # pondération
            score = np.sum(weights * membership)
            scores[c] = score

        pred = max(scores, key=scores.get)
        preds.append(pred)

    return np.array(preds)
```

```python
from sklearn.metrics import accuracy_score

y_pred = fuzzy_predict(X_test_sel, mf, weights)

acc = accuracy_score(y_test, y_pred)
print("QLF Accuracy:", acc)
```

```python
import tensorflow as tf
import cv2
import matplotlib.pyplot as plt

def grad_cam(model, img_array, layer_name):
    grad_model = tf.keras.models.Model(
        [model.inputs],
        [model.get_layer(layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        loss = tf.reduce_max(predictions)

    grads = tape.gradient(loss, conv_outputs)

    weights = tf.reduce_mean(grads, axis=(0, 1, 2))
    cam = tf.reduce_sum(tf.multiply(weights, conv_outputs), axis=-1)

    cam = np.maximum(cam, 0)
    cam = cam / np.max(cam)

    return cam
```
